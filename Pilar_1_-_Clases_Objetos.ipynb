```python
# PILAR 1: CLASES Y OBJETOS
# Autor: Eduardo Zambrano
# Este programa demuestra el concepto de Clases y Objetos en POO
# Una clase es como un "molde" estático, los objetos son instancias en memoria

# DECLARACIÓN DE LA CLASE
# ========================
# Definimos la clase Estudiante que servirá como plantilla para crear objetos
# Declaro la clase, las variables y sus atributos
class Estudiante:
# ATRIBUTOS DE CLASE (COMPARTIDOS POR TODOS LOS OBJETOS)
# ------------------------------------------------------
# Estos atributos pertenecen a la clase, no a cada objeto individual
# Son como "variables estáticas" - ocupan un solo espacio en memoria
    NOTA_MAXIMA = 100  # Constante que define la nota máxima posible
    PORCENTAJE_APROBACION = 0.65  # 65% es el mínimo para aprobar
    estudiantes = []# LISTA ESTÁTICA: Aquí guardaremos TODOS los estudiantes creados

# MÉTODO CONSTRUCTOR __init__
# ============================
# Este método se ejecuta AUTOMÁTICAMENTE cuando creamos un nuevo objeto
# Cada vez que se llama, crea un nuevo ESPACIO EN MEMORIA para un estudiante
    def __init__(self, nombre, edad, carrera, nota):
# ATRIBUTOS DE INSTANCIA (CAMBIAN POR CADA OBJETO)
# ------------------------------------------------
# Estos atributos se almacenan en memoria para CADA objeto individual
# Si creamos 1000 estudiantes, habrá 1000 copias de estos atributos en memoria
        self.nombre = nombre  # Cada estudiante tiene su propio nombre
        self.edad = edad      # Cada estudiante tiene su propia edad
        self.carrera = carrera  # Cada estudiante tiene su propia carrera
        self.nota = nota      # Cada estudiante tiene su propia nota

# El atributo resultado se calcula INTERNAMENTE llamando a un método privado
# El usuario no necesita pasar este dato, la clase lo calcula sola
        self.resultado = self._calcular_resultado(nota)

# IMPORTANTE: Cada vez que se crea un estudiante, se añade a la lista ESTÁTICA
# La lista 'estudiantes' es compartida por TODOS los objetos
        Estudiante.estudiantes.append(self)
# Si creamos 1000 estudiantes, la lista tendrá 1000 referencias, pero
# la lista en sí es un solo objeto en memoria (pertenece a la clase)

# MÉTODO PRIVADO (POR CONVENCIÓN)
# ================================
# El guión bajo _ indica que es un método "privado" (por convención)
# Solo debería ser usado INTERNAMENTE por la clase, no desde fuera
    def _calcular_resultado(self, nota):
#Método interno para calcular si está aprobado o reprobado
        if nota >= (self.NOTA_MAXIMA * self.PORCENTAJE_APROBACION):
            return "APROBADO"  # Si nota >= 65, aprueba
        else:
            return "REPROBADO"  # Si nota < 65, reprueba


    # MÉTODO DE INSTANCIA
    # ===================
    # Este método trabaja con los datos de UNA instancia específica
    # Cada objeto puede llamar a este método y mostrará su propia información
    def mostrar_info(self):
        """Retorna información formateada del estudiante como una fila de tabla"""
        # Los :<20 , :>3 , etc. son para formatear como tabla alineada
        # <20 = alinear a la izquierda con 20 espacios
        # >3 = alinear a la derecha con 3 espacios
        # 6.2f = número con 6 dígitos totales y 2 decimales
        return f"{self.nombre:<20} | {self.edad:>3} | {self.carrera:<25} | {self.nota:>6.2f} | {self.resultado:<10}"


    # MÉTODO DE INSTANCIA
    # ===================
    def presentarse(self):
        """Método simple que retorna una presentación del estudiante"""
        return f"Hola, soy {self.nombre} y estoy estudiando {self.carrera}"

    # MÉTODO DE CLASE (@classmethod)
    # ===============================
    # Los métodos de clase trabajan con la clase en sí, no con instancias
    # Se decoran con @classmethod y reciben 'cls' en lugar de 'self'
    # Son útiles para operaciones que involucran a TODOS los objetos o a la clase

    @classmethod
    def obtener_por_indice(cls, indice):
        """Obtiene un estudiante por su índice (0-based)"""
        if 0 <= indice < len(cls.estudiantes):
            return cls.estudiantes[indice]
        else:
            return None

    @classmethod
    def mostrar_estudiante(cls, indice):
        """Muestra un estudiante específico por su índice"""
        estudiante = cls.obtener_por_indice(indice)

        if estudiante:
            print("\n" + "="*60)
            print("ESTUDIANTE SELECCIONADO")
            print("="*60)
            print(f"{'NOMBRE':<20} | {'EDAD':>3} | {'CARRERA':<25} | {'NOTA':>6} | {'RESULTADO':<10}")
            print("-"*60)
            print(estudiante.mostrar_info())
            print("="*60)
        else:
            print(f"No existe estudiante en el índice {indice}")



    @classmethod
    def menu_seleccion(cls):
        """Muestra un menú para seleccionar estudiante"""
        while True:
            print("\n" + "="*60)
            print("SELECCIONAR ESTUDIANTE")
            print("="*60)

            # Mostrar lista con números
            for i, estudiante in enumerate(cls.estudiantes):
                print(f"{i}. {estudiante.nombre} - {estudiante.carrera}")

            print("\n0. Salir")

            try:
                opcion = int(input("\nSeleccione un número: "))

                if opcion == 0:
                    print("¡Hasta luego!")
                    break

                if 1 <= opcion <= len(cls.estudiantes):
                    cls.mostrar_estudiante(opcion - 1)# (Opcion-1)
                    # Debido al inicio en 0 del indice, el usuario Cuenta los
                    # estudiantes desde 1
                else:
                    print("Opción no válida")

            except ValueError:
                print("Por favor, ingrese un número válido")


# CREACIÓN DE OBJETOS (INSTANCIAS)
# ================================
# Aquí estamos CREANDO objetos (instancias) de la clase Estudiante
# CADA VEZ que llamamos a Estudiante() se crea un NUEVO espacio en memoria
# Creando la "base de datos" para los registros de los estudiantes
#En un programa esto no va en el codigo,
#se tendria una base de datos implementada de otra forma.

# Creación de 5 objetos estudiante
# Cada uno tiene su PROPIO espacio en memoria con sus propios atributos
Estudiante("Ana López García", 20, "Ingeniería Informática", 85),
Estudiante("Carlos Ruiz", 22, "Medicina", 60),
Estudiante("María González Pérez", 21, "Derecho", 70),
Estudiante("José Hernández", 23, "Arquitectura", 95),
Estudiante("Laura Martínez", 19, "Diseño Gráfico", 45),


# NOTA: Después de crear estos 5 estudiantes, la lista Estudiante.estudiantes
# contiene 5 referencias a objetos distintos en memoria

# DEMOSTRACIÓN DE IDs EN MEMORIA
# ==============================
# Cada objeto tiene un ID ÚNICO en memoria, demostrando que son independientes
print("\n" + "="*75)
print("DEMOSTRACIÓN DE IDs EN MEMORIA (CADA OBJETO ES ÚNICO)")
print("="*75)
print("Los IDs diferentes demuestran que cada objeto ocupa su propio espacio")
print("-"*75)

# Iteramos sobre la lista de estudiantes para mostrar sus IDs
for i, estudiante in enumerate(Estudiante.estudiantes):
    # La función id() retorna el identificador único en memoria de cada objeto
    print(f"Estudiante {i}: {estudiante.nombre:<20} | ID en memoria: {id(estudiante)}")

print("\n✅ CONCLUSIÓN: Cada estudiante tiene un ID diferente, son objetos INDEPENDIENTES")
print("\n")

# DEMOSTRACIÓN DE MÉTODOS DE INSTANCIA
# ====================================
# Cada objeto responde de manera diferente a los mismos métodos
print("\n" + "="*75)
print("DEMOSTRACIÓN DE MÉTODOS DE INSTANCIA")
print("="*75)
print("Cada objeto responde con sus propios datos\n")

# Probamos el método presentarse() con diferentes estudiantes
print("--- Método presentarse() ---")
print(Estudiante.estudiantes[0].presentarse())  # Ana se presenta
print(Estudiante.estudiantes[1].presentarse())  # Carlos se presenta
print(Estudiante.estudiantes[2].presentarse())  # María se presenta

print("\n✅ CONCLUSIÓN: Cada Objeto puede responder usando sus propios atributos, son usados de forma INDEPENDIENTES, a pesar de ser partes del mismo objeto")
print("\n")
print("#"*75)
# DEMOSTRACIÓN DE ATRIBUTO CALCULADO INTERNAMENTE
# ===============================================
print("\n--- Atributo 'resultado' calculado automáticamente ---")
print("El programa evalúa si la nota está por encima del 65%")
print("-" * 50)

for estudiante in Estudiante.estudiantes:
    print(f"{estudiante.nombre:<20} | Nota {estudiante.nota} → {estudiante.resultado}")

print("\n")
print("#"*75)
print("==========MUESTRA LOS RESULTADOS DE TODOS ESTUDIANTES==========")
print("\n")
# Imprimir encabezado
print("=" * 80)
print(f"{'Indice':>0}|{'NOMBRE':<16} | {'EDAD':>3} | {'CARRERA':<24} | {'NOTA':>6} | {'RESULTADO':<10}")
print("-" * 80)
# Imprimir cada estudiante
for i, estudiante in enumerate(Estudiante.estudiantes):
    print(f"{i}: {estudiante.mostrar_info()}")
print("=" * 75)
print("\n")
print("#"*80)
# DEMOSTRACIÓN DE ACCESO A ESTUDIANTES ESPECÍFICOS
# ================================================
print("\n" + "="*75)
print("DEMOSTRACIÓN DE ACCESO A ESTUDIANTES ESPECÍFICOS")
print("="*75)

# Usando el método de clase para mostrar un estudiante por índice
print("\n--- Mostrando estudiante en índice 2 (María) ---")
Estudiante.mostrar_estudiante(2)

print("\n--- Mostrando estudiante en índice 0 (Ana) ---")
Estudiante.mostrar_estudiante(0)

print("\n--- Intentando mostrar índice inexistente (10) ---")
Estudiante.mostrar_estudiante(10)  # Esto dará mensaje de error controlado

Estudiante.menu_seleccion()
```
