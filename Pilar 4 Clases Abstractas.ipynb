```python
# Pilar4_Clases_Abstractas_DEMO_ERROR.ipynb
# Autor: Eduardo Zambrano
# Versión: DEMOSTRACIÓN PRÁCTICA DE ERROR EN TIEMPO DE EJECUCIÓN

from abc import ABC, abstractmethod
import random
import os
import time

# ============================================================================
# CLASE ABSTRACTA: Define el CONTRATO
# ============================================================================

class Personaje(ABC):
    """
    Una Clase ABSTRACTA: Es como un "molde" o "contrato" que dice:
    "Todo personaje en mi juego DEBE poder atacar y defender"
    """

    def __init__(self, nombre, nivel=1):
        """
        Atributos que TODOS los personajes tienen
        """
        self.nombre = nombre
        self.nivel = nivel
        self.vida = 100 + (nivel * 10)
        self.vida_maxima = self.vida
        self.experiencia = 0
        print(f"✅ {self.nombre} está siendo creado...")

    @abstractmethod
    def atacar(self):
        """MÉTODO ABSTRACTO: Todo personaje DEBE implementar su forma de atacar"""
        pass

    @abstractmethod
    def defender(self, dano):
        """MÉTODO ABSTRACTO: Todo personaje DEBE implementar su forma de defender"""
        pass

    def recibir_dano(self, dano):
        """MÉTODO CONCRETO: Todos reciben daño IGUAL"""
        self.vida -= dano
        if self.vida < 0:
            self.vida = 0
        return dano

    def esta_vivo(self):
        return self.vida > 0

    def mostrar_estado(self):
        barra = "█" * int(self.vida / 10) + "░" * int((self.vida_maxima - self.vida) / 10)
        return f"{self.nombre} [{barra}] {self.vida}/{self.vida_maxima} HP (Nv.{self.nivel})"

    def ganar_experiencia(self, cantidad):
        self.experiencia += cantidad
        if self.experiencia >= 100:
            self.nivel += 1
            self.experiencia = 0
            self.vida_maxima += 20
            self.vida = self.vida_maxima
            return True
        return False

    def saludar(self):
        return f"¡Hola! Soy {self.nombre}"


# ============================================================================
# CLASES COMPLETAS (implementan TODO el contrato)
# ============================================================================

class Guerrero(Personaje):
    """Implementa CORRECTAMENTE atacar() y defender()"""

    def __init__(self, nombre, nivel=1):
        super().__init__(nombre, nivel)
        self.tipo = "GUERRERO ⚔️"
        self.fuerza = 15 + (nivel * 2)

    def atacar(self):
        probabilidad_acierto = 0.8
        if random.random() < probabilidad_acierto:
            dano = self.fuerza + random.randint(5, 15)
            print(f"⚔️ {self.nombre} ataca con su ESPADA causando {dano} de daño!")
            return dano
        else:
            print(f"⚔️ {self.nombre} falló el ataque con su espada!")
            return 0

    def defender(self, dano):
        reduccion = random.randint(5, 15)
        dano_real = max(0, dano - reduccion)
        print(f"🛡️ {self.nombre} bloquea con su ESCUDO! Reduce {reduccion} de daño.")
        return self.recibir_dano(dano_real)


class Mago(Personaje):
    """Implementa CORRECTAMENTE atacar() y defender()"""

    def __init__(self, nombre, nivel=1):
        super().__init__(nombre, nivel)
        self.tipo = "MAGO 🔮"
        self.mana = 50 + (nivel * 5)
        self.inteligencia = 20 + (nivel * 3)

    def atacar(self):
        if self.mana >= 10:
            self.mana -= 10
            dano = self.inteligencia + random.randint(10, 25)
            print(f"🔮 {self.nombre} lanza un HECHIZO causando {dano} de daño! (Mana: {self.mana})")
            return dano
        else:
            print(f"🔮 {self.nombre} no tiene mana! Solo hace 5 de daño.")
            return 5

    def defender(self, dano):
        if self.mana >= 15:
            self.mana -= 15
            reduccion = random.randint(10, 25)
            dano_real = max(0, dano - reduccion)
            print(f"✨ {self.nombre} crea un CAMPO DE FUERZA! Reduce {reduccion} de daño.")
            return self.recibir_dano(dano_real)
        else:
            print(f"✨ {self.nombre} no tiene mana para defender! Recibe daño completo.")
            return self.recibir_dano(dano)


class Arquero(Personaje):
    """Implementa CORRECTAMENTE atacar() y defender()"""

    def __init__(self, nombre, nivel=1):
        super().__init__(nombre, nivel)
        self.tipo = "ARQUERO 🏹"
        self.agilidad = 18 + (nivel * 2)
        self.punteria = 0.9

    def atacar(self):
        if random.random() < self.punteria:
            dano = self.agilidad + random.randint(8, 18)
            print(f"🏹 {self.nombre} dispara una FLECHA causando {dano} de daño!")
            return dano
        else:
            print(f"🏹 {self.nombre} falló el disparo!")
            return 0

    def defender(self, dano):
        probabilidad_esquiva = 0.3
        if random.random() < probabilidad_esquiva:
            print(f"💨 {self.nombre} ESQUIVA el ataque completamente!")
            return self.recibir_dano(0)
        else:
            reduccion = random.randint(3, 10)
            dano_real = max(0, dano - reduccion)
            print(f"💨 {self.nombre} intenta esquivar, pero recibe {dano_real} de daño.")
            return self.recibir_dano(dano_real)


# ============================================================================
# CLASES INCOMPLETAS (NO implementan el contrato) - ¡ESTO CAUSARÁ ERROR!
# ============================================================================

class Curandero(Personaje):
    """
    ⚠️  CLASE INCOMPLETA: No implementa atacar() ni defender()
    """
    def __init__(self, nombre, nivel=1):
        super().__init__(nombre, nivel)
        self.tipo = "CURANDERO 💊"
    # ❌ FALTA: atacar()
    # ❌ FALTA: defender()


class Demihumano(Personaje):
    """
    ⚠️  CLASE INCOMPLETA: Solo implementa UNO de los dos métodos
    """
    def __init__(self, nombre, nivel=1):
        super().__init__(nombre, nivel)
        self.tipo = "DEMIHUMANO 🧝"

    def atacar(self):
        print(f"🧝 {self.nombre} ataca con GARRA!")
        return random.randint(10, 20)
    # ❌ FALTA: defender()


# ============================================================================
# FUNCIONES DE UTILIDAD
# ============================================================================

def limpiar_pantalla():
    os.system('cls' if os.name == 'nt' else 'clear')


def combatir(jugador1, jugador2):
    """Simula un combate entre dos personajes"""
    print("\n" + "="*70)
    print(f"⚔️  COMBATE: {jugador1.nombre} vs {jugador2.nombre} ⚔️")
    print("="*70)

    ronda = 1
    while jugador1.esta_vivo() and jugador2.esta_vivo():
        print(f"\n--- RONDA {ronda} ---")

        # Jugador 1 ataca a Jugador 2
        print(f"\n▶️ Turno de {jugador1.nombre}:")
        dano = jugador1.atacar()
        if dano > 0:
            jugador2.defender(dano)

        if not jugador2.esta_vivo():
            print(f"\n💀 ¡{jugador2.nombre} ha sido derrotado!")
            break

        # Jugador 2 ataca a Jugador 1
        print(f"\n▶️ Turno de {jugador2.nombre}:")
        dano = jugador2.atacar()
        if dano > 0:
            jugador1.defender(dano)

        if not jugador1.esta_vivo():
            print(f"\n💀 ¡{jugador1.nombre} ha sido derrotado!")
            break

        print(f"\n📊 Estado después de ronda {ronda}:")
        print(jugador1.mostrar_estado())
        print(jugador2.mostrar_estado())

        ronda += 1
        time.sleep(1)

    if jugador1.esta_vivo():
        print(f"\n🏆 ¡{jugador1.nombre} GANA el combate!")
        jugador1.ganar_experiencia(50)
        jugador2.ganar_experiencia(20)
    else:
        print(f"\n🏆 ¡{jugador2.nombre} GANA el combate!")
        jugador2.ganar_experiencia(50)
        jugador1.ganar_experiencia(20)


def crear_personaje_normal():
    """Crea un personaje de los que FUNCIONAN (Guerrero, Mago, Arquero)"""
    print("\n" + "="*60)
    print("👤 CREAR NUEVO PERSONAJE")
    print("="*60)

    print("\nElige tu clase:")
    print("1. ⚔️ Guerrero")
    print("2. 🔮 Mago")
    print("3. 🏹 Arquero")

    try:
        opcion = input("\nSelecciona clase (1-3): ")
        nombre = input("Nombre de tu personaje: ")
        nivel = int(input("Nivel de tu personaje: "))

        if opcion == '1':
            return Guerrero(nombre, nivel)
        elif opcion == '2':
            return Mago(nombre, nivel)
        elif opcion == '3':
            return Arquero(nombre, nivel)
        else:
            print("❌ Opción no válida")
            return None
    except:
        print("❌ Error al crear personaje")
        return None


def crear_personaje_demo(tipo, nombre, nivel):
    """Intenta crear un personaje para la DEMO (puede fallar)"""
    try:
        if tipo == "1":
            return True, Guerrero(nombre, nivel)
        elif tipo == "2":
            return True, Mago(nombre, nivel)
        elif tipo == "3":
            return True, Arquero(nombre, nivel)
        elif tipo == "4":
            # ⚠️  ESTO VA A FALLAR
            return True, Curandero(nombre, nivel)
        elif tipo == "5":
            # ⚠️  ESTO VA A FALLAR
            return True, Demihumano(nombre, nivel)
        elif tipo == "6":
            return False, "SALIR"
        else:
            return False, "Opción no válida"

    except TypeError as e:
        return False, f"❌ ERROR DE ABSTRACCIÓN: {e}"
    except Exception as e:
        return False, f"❌ Error inesperado: {e}"


def mostrar_personajes(lista):
    """Muestra todos los personajes creados"""
    if not lista:
        print("\n📭 No hay personajes creados aún")
        return

    print("\n" + "="*70)
    print("📋 PERSONAJES DISPONIBLES")
    print("="*70)
    for i, p in enumerate(lista, 1):
        print(f"{i}. {p.tipo} - {p.mostrar_estado()}")


# ============================================================================
# MENÚ DE DEMOSTRACIÓN DE ERRORES
# ============================================================================

def menu_demo():
    """Menú que muestra QUÉ PASA con clases incompletas"""

    print("\n" + "="*80)
    print("🧪  DEMOSTRACIÓN DE CLASES ABSTRACTAS - ERRORES EN TIEMPO REAL")
    print("="*80)
    print("""
    📚 EXPLICACIÓN:

    Una CLASE ABSTRACTA: Es un ideario, principio o concepto que regula el
    funcionamiento de una clase. Es como un CONTRATO que dice en este caso:
    "TODO personaje (Clase) DEBE tener ataque y defensa (Atributos)"

    ✅ COMPLETAS (FUNCIONAN):
    ❌ INCOMPLETAS (FALLAN):
    Guerrero, Mago, Arquero, Curandero, Demihumano

    🔴 Cuando intentes crear un Personaje que no cumple con el contrato,
     PYTHON DARÁ ERROR y se encargara de mostrartelo. podemos "capturar"
     el error para identificar que falta en la clase que construimos
     como se ve en el ejeplo.
    """)

    input("\nPresiona ENTER para ver la demostración...")

    while True:
        limpiar_pantalla()

        print("\n" + "="*80)
        print("🎮  DEMOSTRACIÓN - CREACIÓN DE PERSONAJES")
        print("="*80)
        print("""
        ✅ COMPLETOS (FUNCIONAN):
        1. ⚔️  Guerrero
        2. 🔮 Mago
        3. 🏹 Arquero
        4. 💊 Curandero
        5. 🧝 Demihumano
        ❌ LOS PERSONAJES "INCOMPLETOS" VAN A FALLAR EN SU CREACION

        6. ❌ VOLVER AL MENÚ PRINCIPAL
        """)
        print("-"*80)

        opcion = input("🔷 ¿Qué personaje quieres crear? (1-5): ")

        if opcion == '6':
            break

        nombre = input("📝 Nombre del personaje: ")
        nivel = int(input("🔢 Nivel del personaje: "))

        # Intentamos crear el personaje
        exito, resultado = crear_personaje_demo(opcion, nombre)

        print("\n" + "-"*80)
        print("🔍 RESULTADO:")
        print("-"*80)
        print(f"📋 Evaluemos al Personaje creado ataquemoslo")
        if exito:
            print(f"✅ ¡PERSONAJE CREADO EXITOSAMENTE!")
            print(f"\n   {resultado.saludar()}")
            # Probamos sus métodos
            print(f"   {resultado.atacar()}")
            # Para defender necesitamos un valor de prueba
            if hasattr(resultado, 'defender'):
                resultado.defender(10)
            print(f"\n  EXITO!  {resultado.mostrar_estado()} El personaje cumple con lo necesario para combatir en el juego")
        else:
            print(f"\n{resultado}")
            if "ERROR DE ABSTRACCIÓN" in str(resultado):
                print("\n   ⚠️  FALLO CATASTROFICO, EL PERSONAJE NO TIENE DEFENSA:")
                print("\n   ⚠️  ESTE ES EL ERROR DE ABSTRACTACIÓN:")
                print("   Python NO permite crear objetos de clases")
                print("   que no implementen TODOS los métodos abstractos.")

        print("\n" + "="*80)
        input("Presiona ENTER para continuar...")


# ============================================================================
# MENÚ DE EXPLICACIÓN TÉCNICA
# ============================================================================

def mostrar_explicacion():
    """Muestra la explicación técnica"""
    limpiar_pantalla()

    print("\n" + "="*80)
    print("📚 EXPLICACIÓN TÉCNICA - CLASES ABSTRACTAS")
    print("="*80)
    print("""
    ¿QUÉ ES UNA CLASE ABSTRACTA?
    ----------------------------------------
    Es una clase que NO PUEDE ser instanciada directamente.
    Sirve como PLANTILLA o CONTRATO para otras clases.

    🔴 ¿QUÉ PASA SI INTENTAS INSTANCIAR UNA CLASE ABSTRACTA?
    ----------------------------------------
    Python lanza: TypeError: Can't instantiate abstract class
    Ejemplo: "TypeError: Can't instantiate abstract class Curandero
              with abstract methods atacar, defender"

    🔴 ¿QUÉ PASA SI UNA CLASE HIJA NO IMPLEMENTA TODO?
    ----------------------------------------
    TAMPOCO se puede instanciar. Python te dice QUÉ métodos faltan.

    ✅ ¿PARA QUÉ SIRVE?
    ----------------------------------------
    1. GARANTIZA que todas las clases hijas tengan cierto comportamiento
    2. Es como un "checklist" que Python revisa automáticamente
    3. Evita errores en tiempo de ejecución
    """)

    input("\nPresiona ENTER para volver...")


# ============================================================================
# MENÚ DE JUEGO (con los personajes que FUNCIONAN)
# ============================================================================

def menu_juego():
    """Menú del juego REAL con personajes que funcionan"""

    # Lista de personajes disponibles
    personajes = [
        Guerrero("Guts", nivel=3),
        Mago("Gandalf", nivel=2),
        Arquero("Legolas", nivel=2)
    ]

    while True:
        limpiar_pantalla()

        print("\n" + "="*70)
        print("🎮  SISTEMA DE PERSONAJES - MENÚ PRINCIPAL")
        print("="*70)
        print(f"👥 Personajes disponibles: {len(personajes)}")
        print("-"*70)
        print("1. 👤 CREAR NUEVO PERSONAJE")
        print("2. 📋 VER PERSONAJES")
        print("3. ⚔️  COMBATIR")
        print("4. 📚 EXPLICACIÓN")
        print("5. 🔙 VOLVER")
        print("-"*70)

        opcion = input("🔷 Selecciona una opción: ")

        if opcion == '1':
            nuevo = crear_personaje_normal()
            if nuevo:
                personajes.append(nuevo)
                print(f"\n✅ ¡{nuevo.nombre} ha sido creado!")
            input("\nPresiona ENTER...")

        elif opcion == '2':
            mostrar_personajes(personajes)
            input("\nPresiona ENTER...")

        elif opcion == '3':
            if len(personajes) < 2:
                print("\n❌ Necesitas al menos 2 personajes para combatir")
                input("Presiona ENTER...")
                continue

            mostrar_personajes(personajes)
            try:
                p1 = int(input("\nSelecciona primer personaje: ")) - 1
                p2 = int(input("Selecciona segundo personaje: ")) - 1

                if p1 == p2:
                    print("❌ No puedes combatir contra ti mismo")
                elif 0 <= p1 < len(personajes) and 0 <= p2 < len(personajes):
                    combatir(personajes[p1], personajes[p2])
                else:
                    print("❌ Selección no válida")
            except:
                print("❌ Error en la selección")

            input("\nPresiona ENTER...")

        elif opcion == '4':
            mostrar_explicacion()

        elif opcion == '5':
            break


# ============================================================================
# PROGRAMA PRINCIPAL
# ============================================================================

def main():
    while True:
        limpiar_pantalla()

        print("\n" + "="*80)
        print("🧪  LABORATORIO DE CLASES ABSTRACTAS")
        print("="*80)
        print("""
        1. 🎮 JUGAR (personajes que FUNCIONAN)
        2. 🧪 DEMOSTRACIÓN DE ERRORES (verás los fallos)
        3. 📚 EXPLICACIÓN TÉCNICA
        4. ❌ SALIR
        """)
        print("-"*80)

        opcion = input("🔷 Selecciona: ")

        if opcion == '1':
            menu_juego()
        elif opcion == '2':
            menu_demo()
        elif opcion == '3':
            mostrar_explicacion()
        elif opcion == '4':
            print("\n👋 ¡Gracias por Jugar! ¡Hasta luego!")
            break
        else:
            print("❌ Opción no válida")
            input("Presiona ENTER...")


if __name__ == "__main__":
    main()
```
