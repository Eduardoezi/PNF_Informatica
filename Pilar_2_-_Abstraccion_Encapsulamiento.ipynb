```python
# Pilar2 Abstraccion y Encapsulamiento.ipynb
# Autor: Eduardo Zambrano
# Este programa demuestra los conceptos de Abstracción y Encapsulamiento

import os  # Para limpiar pantalla
import random  # Para generar número de cuenta

class CuentaBancaria:
    """
    Una clase que representa una cuenta bancaria con encapsulamiento.

    ENCAPSULAMIENTO: Ocultar los detalles internos y proteger los datos.
    En Python, esto se logra por CONVENCIÓN (no es obligatorio como en otros lenguajes)
    """

    def __init__(self, titular, saldo_inicial, codigo_banco="0105"):
        """
        Constructor de la clase.

        self.titular: ATRIBUTO PÚBLICO - Cualquiera puede verlo y modificarlo directamente
        self._saldo: ATRIBUTO PROTEGIDO (por convención) - Un guión bajo indica "no tocar directamente"
        self.__numero_cuenta: ATRIBUTO PRIVADO - Doble guión activa "name mangling" (ofuscación)
        """
        self.titular = titular  # Público: accesible desde cualquier parte

        # PROTEGIDO: Convención dice que no deberías acceder directamente
        # Pero Python no lo impide, solo es una "advertencia" al programador
        self._saldo = saldo_inicial

        self.codigo_banco = codigo_banco  # Guardamos el código del banco

        # PRIVADO: Python "oculta" el nombre de la variable
        # Internamente lo renombra a _CuentaBancaria__numero_cuenta
        self.__numero_cuenta = self._generar_numero_cuenta()

    def _generar_numero_cuenta(self):
        """
        Genera el número COMPLETO (3 bloques de 4 dígitos)
        El primer bloque es FIJO (identifica al banco)
        Los otros dos son aleatorios
        """
        # Bloque 1: Código del banco (fijo, como string para mantener el 0 inicial)
        bloque1 = self.codigo_banco  # Ejemplo: "0105"

        # Bloques aleatorios
        bloque2 = str(random.randint(1000, 9999))
        bloque3 = str(random.randint(1000, 9999))
        bloque4 = str(random.randint(1000, 9999))

        # Formato final: ES010512345678
        return f"ES{bloque1}{bloque2}{bloque3}{bloque4}"

    @property
    def saldo(self):
        """
        GETTER - PROPERTY
        Un getter permite CONTROLAR CÓMO se accede a un atributo privado/protegido

        Ventajas:
        1. Podemos formatear la salida (añadir símbolo $, comas, etc.)
        2. Podemos añadir lógica adicional sin cambiar la interfaz
        3. Mantenemos el control aunque parezca que accedemos directamente
        """
        return f"${self._saldo:,.2f}"  # Formato: $1,000.00

    @property
    def saldo_numerico(self):
        """Getter que devuelve el número REAL para operaciones"""
        return self._saldo

    @property
    def numero_cuenta_publico(self):
        """
        Versión PÚBLICA: Mostramos el código del banco pero ocultamos el resto
        Ejemplo: "0105-****-****"
        """
        # Mostramos código del banco y últimos 6 dígitos
        return f"{self.codigo_banco}-****-****-**{self.__numero_cuenta[-6:]}"

    @property
    def numero_cuenta_completo(self):
        """Número completo (solo para uso interno)"""
        return self.__numero_cuenta

    def mostrar_numero_completo(self):
        """Método especial para ver el número real"""
        print("\n" + "🔐"*30)
        print("INFORMACIÓN COMPLETA DE LA CUENTA")
        print("🔐"*30)
        print(f"Número COMPLETO: {self.__numero_cuenta}")
        print("🔐"*30)

    def mismo_banco(self, otra_cuenta):
        """Verifica si otra cuenta es del mismo banco"""
        return self.codigo_banco == otra_cuenta.codigo_banco

    @saldo.setter
    def saldo(self, nuevo_saldo):
        """
        SETTER
        Permite CONTROLAR CÓMO se MODIFICA un atributo

        Aquí podemos:
        1. Validar que el nuevo valor sea válido
        2. Añadir advertencias
        3. Registrar cambios, etc.
        """
        # Validación 1: No permitir saldos negativos
        if nuevo_saldo < 0:
            raise ValueError("El saldo no puede ser negativo")

        # Validación 2: Advertencia para depósitos grandes
        if nuevo_saldo > 1000000:
            print("⚠️  Advertencia: Depósito muy grande, se requerirá verificación")

        # Si pasa las validaciones, actualizamos el valor
        self._saldo = nuevo_saldo

    def depositar(self, cantidad):
        """
        Método público para depositar dinero
        Encapsula la lógica de negocio: validar que la cantidad sea positiva
        """
        if cantidad > 0:
            self._saldo += cantidad
            return True, f"✅ Depósito exitoso. Nuevo saldo: {self.saldo}"
        else:
            return False, "❌ La cantidad a depositar debe ser positiva"

    def retirar(self, cantidad):
        """
        Método público para retirar dinero
        Encapsula la lógica de negocio: validar saldo suficiente
        """
        if cantidad <= 0:
            return False, "❌ La cantidad debe ser positiva"
        elif cantidad > self._saldo:
            return False, f"❌ Saldo insuficiente. Disponible: {self.saldo}"
        else:
            self._saldo -= cantidad
            return True, f"✅ Retiro exitoso. Nuevo saldo: {self.saldo}"


# Funciones fuera de la clase.
def limpiar_pantalla():
    """Función para limpiar la pantalla (funciona en Windows y Linux/Mac)"""
    os.system('cls' if os.name == 'nt' else 'clear')


def mostrar_titular(cuenta):
    """Muestra la información principal de la cuenta"""
    print("\n" + "="*60)
    print("🏦  CAJERO AUTOMÁTICO - BANCO POO - CÓDIGO 0105")
    print("="*60)
    print(f"👤 Titular: {cuenta.titular}")
    print(f"🔢 Cuenta: {cuenta.numero_cuenta_publico}")
    print(f"💰 Saldo: {cuenta.saldo}")
    print("-"*60)


def menu_principal():
    """Muestra el menú de opciones"""
    print("\n📋 OPERACIONES DISPONIBLES:")
    print("   1. 💰 DEPOSITAR dinero")
    print("   2. 💸 RETIRAR dinero")
    print("   3. 🔢 VER NÚMERO COMPLETO (demostración)")
    print("   4. 👤 CREAR OTRA CUENTA DEL MISMO BANCO")
    print("   5. 🔄 CAMBIAR DE CLIENTE")
    print("   6. ❌ SALIR")
    print("-"*60)


def confirmar_operacion(tipo, cantidad, cuenta):
    """Pide confirmación antes de realizar la operación"""
    print(f"\n⚡ CONFIRMACIÓN REQUERIDA:")
    print(f"   Operación: {tipo}")
    print(f"   Cantidad: ${cantidad:,.2f}")
    print(f"   Saldo actual: {cuenta.saldo}")

    respuesta = input("\n¿Confirmar operación? (s/n): ").lower()
    return respuesta == 's' or respuesta == 'si'


def menu_seleccion_cliente(clientes):
    """
    Muestra el menú para seleccionar un cliente
    Recibe la lista de clientes y retorna el índice seleccionado
    """
    while True:
        limpiar_pantalla()
        print("\n" + "="*60)
        print("🏦  BANCO POO - SELECCIÓN DE CLIENTE")
        print("="*60)
        print("\n📋 CLIENTES DISPONIBLES:")
        print("-"*40)

        # Mostrar lista de clientes con números
        for i, cliente in enumerate(clientes):
            print(f"   {i+1}. {cliente.titular} - Saldo: {cliente.saldo}")

        print("-"*40)
        print("   0. 👤 CREAR NUEVO CLIENTE")
        print("="*60)

        try:
            opcion = int(input("\n🔷 Seleccione un cliente (0-{}): ".format(len(clientes))))

            if opcion == 0:
                # Crear nuevo cliente
                return "CREAR_NUEVO"
            elif 1 <= opcion <= len(clientes):
                # Retornar el índice (0-based) del cliente seleccionado
                return opcion - 1
            else:
                print("❌ Opción no válida")
                input("\nPresione ENTER para continuar...")
        except ValueError:
            print("❌ Por favor, ingrese un número válido")
            input("\nPresione ENTER para continuar...")


def crear_nuevo_cliente(clientes):
    """Función para crear un nuevo cliente y agregarlo a la lista"""
    print("\n--- CREAR NUEVO CLIENTE ---")
    nombre = input("Nombre del titular: ")

    try:
        saldo_inicial = float(input("Saldo inicial: $"))

        # Crear la nueva cuenta
        nueva_cuenta = CuentaBancaria(nombre, saldo_inicial, codigo_banco="0105")

        # Agregar a la lista de clientes
        clientes.append(nueva_cuenta)

        print(f"\n✅ Cliente creado exitosamente!")
        print(f"Titular: {nueva_cuenta.titular}")
        print(f"Cuenta: {nueva_cuenta.numero_cuenta_publico}")
        print(f"Número completo: {nueva_cuenta.numero_cuenta_completo}")

        input("\nPresione ENTER para continuar...")

        # Retornar el índice del nuevo cliente (último de la lista)
        return len(clientes) - 1

    except ValueError:
        print("❌ Error: Ingrese un número válido para el saldo")
        input("\nPresione ENTER para continuar...")
        return None


def ejecutar_cajero():
    """Función principal que ejecuta el cajero automático"""

    # =========================================================
    # LISTA DE CLIENTES PRE-CARGADOS (crece con nuevas cuentas)
    # =========================================================
    clientes = [
        CuentaBancaria("Pedro Gómez", 1500.00, codigo_banco="0105"),
        CuentaBancaria("Juan Pérez", 1000.00, codigo_banco="0105"),
        CuentaBancaria("Alberto López", 2500.00, codigo_banco="0105"),
        CuentaBancaria("José Rodríguez", 800.00, codigo_banco="0105"),
        CuentaBancaria("María García", 3200.00, codigo_banco="0105"),
        CuentaBancaria("Isabel Martínez", 1750.00, codigo_banco="0105"),
    ]

    while True:
        # Seleccionar cliente (o crear nuevo)
        seleccion = menu_seleccion_cliente(clientes)

        if seleccion == "CREAR_NUEVO":
            indice_cliente = crear_nuevo_cliente(clientes)
            if indice_cliente is None:
                continue  # Si hubo error, volver al menú de selección
        else:
            indice_cliente = seleccion

        # Obtener la cuenta seleccionada
        cuenta_actual = clientes[indice_cliente]

        # Menú de operaciones para el cliente seleccionado
        while True:
            limpiar_pantalla()
            mostrar_titular(cuenta_actual)
            menu_principal()

            try:
                opcion = input("🔷 Seleccione una opción (1-6): ").strip()

                if opcion == '1':  # DEPOSITAR
                    try:
                        cantidad = float(input("\n💰 Ingrese cantidad a depositar: $"))
                        exito, mensaje = cuenta_actual.depositar(cantidad)
                        print(f"\n{mensaje}")
                        input("\nPresione ENTER...")
                    except ValueError:
                        print("\n❌ Error: Ingrese un número válido")
                        input("\nPresione ENTER...")

                elif opcion == '2':  # RETIRAR
                    try:
                        cantidad = float(input("\n💸 Ingrese cantidad a retirar: $"))

                        print(f"\n   Saldo actual: {cuenta_actual.saldo}")
                        print(f"   Cantidad solicitada: ${cantidad:,.2f}")

                        # Verificación rápida de saldo
                        if cuenta_actual.saldo_numerico - cantidad < 0:
                            print("\n❌ OPERACIÓN RECHAZADA: Saldo insuficiente")
                            print(f"   Saldo disponible: {cuenta_actual.saldo}")
                            input("\nPresione ENTER...")
                            continue

                        if confirmar_operacion("RETIRO", cantidad, cuenta_actual):
                            exito, mensaje = cuenta_actual.retirar(cantidad)
                            print(f"\n{mensaje}")
                        else:
                            print("\n❌ Operación cancelada")

                        input("\nPresione ENTER...")

                    except ValueError:
                        print("\n❌ Error: Ingrese un número válido")
                        input("\nPresione ENTER...")

                elif opcion == '3':  # VER NÚMERO COMPLETO
                    cuenta_actual.mostrar_numero_completo()
                    print(f"\n📍 El cliente solo ve: {cuenta_actual.numero_cuenta_publico}")
                    input("\nPresione ENTER para continuar...")

                elif opcion == '4':  # CREAR NUEVA CUENTA (DESDE EL MENÚ)
                    print("\n--- CREAR NUEVA CUENTA ---")
                    nombre = input("Nombre del titular: ")

                    try:
                        saldo_inicial = float(input("Saldo inicial: $"))
                        nueva_cuenta = CuentaBancaria(nombre, saldo_inicial, codigo_banco="0105")
                        clientes.append(nueva_cuenta)

                        print(f"\n✅ Cuenta creada exitosamente!")
                        print(f"Titular: {nueva_cuenta.titular}")
                        print(f"Cuenta: {nueva_cuenta.numero_cuenta_publico}")

                        # Verificar si es del mismo banco
                        if cuenta_actual.mismo_banco(nueva_cuenta):
                            print(f"✅ Mismo banco (código {cuenta_actual.codigo_banco})")

                        input("\nPresione ENTER...")

                    except ValueError:
                        print("❌ Error: Ingrese un número válido")
                        input("\nPresione ENTER...")

                elif opcion == '5':  # CAMBIAR DE CLIENTE
                    print("\n🔄 Volviendo al selector de clientes...")
                    input("\nPresione ENTER...")
                    break  # Sale del menú actual para volver al selector

                elif opcion == '6':  # SALIR
                    print("\n👋 ¡Gracias por usar el Banco POO!")
                    return  # Termina todo el programa

                else:
                    print("❌ Opción no válida")
                    input("\nPresione ENTER...")

            except KeyboardInterrupt:
                print("\n\n👋 Operación cancelada. ¡Hasta luego!")
                return


# EJECUTAR EL PROGRAMA
if __name__ == "__main__":
    ejecutar_cajero()
```
